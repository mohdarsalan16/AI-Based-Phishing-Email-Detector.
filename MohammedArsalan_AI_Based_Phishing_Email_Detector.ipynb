## Project overview
This beginner-friendly demonstration uses a small, illustrative dataset of email messages and a machine-learning text classifier to predict whether a message looks like **phishing** or **legitimate**. It uses TF-IDF text features and Logistic Regression from scikit-learn.

> **Important limitation:** The included examples are synthetic and intentionally small. This is an educational prototype, not a production security tool. It can make mistakes and must not be used as the only basis for deciding whether an email is safe. Do not paste private or confidential emails into an untrusted environment.


## 1. Install dependencies
If running locally, install the packages listed in `requirements.txt`. In Google Colab or many Jupyter environments, scikit-learn and pandas may already be installed.

```python
# If needed, uncomment and run this line in a notebook environment:
# %pip install -r requirements.txt

import pandas as pd
from IPython.display import display
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print('Libraries imported successfully.')
```

## 2. Create the sample dataset
These short examples are fictional training examples created for this project. The labels are `phishing` and `legitimate`.

```python
examples = [
    ('Urgent: verify your account password immediately using this link', 'phishing'),
    ('Your account will be suspended today unless you confirm your details', 'phishing'),
    ('You have won a prize claim it now by entering your bank information', 'phishing'),
    ('Security alert: sign in here to prevent account closure', 'phishing'),
    ('Please send your one time password to unlock your account', 'phishing'),
    ('Unusual activity detected click the link and enter your credentials', 'phishing'),
    ('Your parcel is waiting pay a small fee to release it now', 'phishing'),
    ('Confirm your payment details immediately to avoid service interruption', 'phishing'),
    ('You are selected for a cash reward provide your card number', 'phishing'),
    ('Final warning: update your login details through this attachment', 'phishing'),
    ('Please review the meeting agenda before our team call tomorrow', 'legitimate'),
    ('Attached is the class timetable for next week', 'legitimate'),
    ('Thank you for attending the project discussion today', 'legitimate'),
    ('The library will close at six pm during the holiday', 'legitimate'),
    ('Your assignment feedback is available on the college portal', 'legitimate'),
    ('Can we reschedule our appointment for Friday afternoon', 'legitimate'),
    ('Here are the notes from yesterday meeting', 'legitimate'),
    ('The monthly newsletter includes updates from our department', 'legitimate'),
    ('Please find the approved event schedule attached', 'legitimate'),
    ('Your registration receipt is available in your account dashboard', 'legitimate'),
    ('The instructor shared the study material with the class', 'legitimate'),
    ('Reminder: our group presentation starts at ten tomorrow', 'legitimate'),
    ('Please confirm whether you can attend the project meeting', 'legitimate'),
    ('The help desk will never ask you to share your password by email', 'legitimate'),
]

df = pd.DataFrame(examples, columns=['email_text', 'label'])
display(df.head())
print(f'Total examples: {len(df)}')
display(df['label'].value_counts().rename('count').to_frame())
```

## 3. Split data and train the model
TF-IDF converts text into numerical features. Logistic Regression learns patterns associated with the labels. A stratified split keeps both classes represented in the test set.

```python
X_train, X_test, y_train, y_test = train_test_split(
    df['email_text'], df['label'], test_size=0.25, random_state=42, stratify=df['label']
)

model = Pipeline([
    ('tfidf', TfidfVectorizer(lowercase=True, stop_words='english', ngram_range=(1, 2))),
    ('classifier', LogisticRegression(max_iter=1000, random_state=42))
])

model.fit(X_train, y_train)
print('Model training complete.')
print(f'Training examples: {len(X_train)} | Test examples: {len(X_test)}')
```

## 4. Evaluate the model
Accuracy alone is not enough for a real security product. Review the per-class precision and recall, and remember that this test set is tiny and synthetic.

```python
predictions = model.predict(X_test)
print(f'Accuracy on this small test split: {accuracy_score(y_test, predictions):.2f}')
print('\nClassification report:')
print(classification_report(y_test, predictions, zero_division=0))
print('Confusion matrix (label order: legitimate, phishing):')
print(confusion_matrix(y_test, predictions, labels=['legitimate', 'phishing']))
```

## 5. Try a sample email
Enter a fictional or non-sensitive message below. The model predicts a label; it does not verify sender identity, inspect attachments, or open links.

```python
def predict_email(email_text):
    """Classify a message with the trained educational model."""
    if not isinstance(email_text, str) or not email_text.strip():
        raise ValueError('Please enter a non-empty email message.')
    label = model.predict([email_text])[0]
    probabilities = model.predict_proba([email_text])[0]
    class_names = model.named_steps['classifier'].classes_
    confidence = float(probabilities[list(class_names).index(label)])
    return {'prediction': label, 'model_score': round(confidence, 3)}

sample_email = 'Urgent: verify your account password immediately using this link'
print(predict_email(sample_email))
```

```python
# Change this example to test another fictional message.
test_email = 'Please review the meeting agenda before our team call tomorrow'
result = predict_email(test_email)
print('Prediction:', result['prediction'])
print('Model score (not a guarantee of safety):', result['model_score'])
```

## 6. Findings, limitations, and future improvements

- The notebook demonstrates a complete basic text-classification workflow.
- The sample dataset is small, fictional, and not representative of real email traffic.
- High or low model scores do not guarantee that an email is safe or malicious.
- A better version would use a larger, ethically sourced and properly labelled dataset, separate validation data, stronger evaluation, and privacy/security review.
- Future work could include URL feature analysis without visiting links, sender-domain checks using trusted data, and a simple user interface.
